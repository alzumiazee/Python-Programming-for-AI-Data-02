## Install Git in windows

### Step 1: Download the Installer

Go to the [⁠Git for Windows Downloads Page](https://git-scm.com/install/windows) to automatically grab the 64-bit standalone installer. Alternatively, if you are using Windows 10 or 11, you can install it instantly via your PowerShell with the following command:```winget install --id Git. Git -e --source winge```

### Step 2: Run the Setup Wizard

1. Open the downloaded ```.exe``` file from your Downloads folder.
2. Click **Next** through the license agreements and destination selections.
3. Stick to the default recommended configurations (like using the Git Credential Manager and bundled OpenSSH) unless you have specific advanced preferences.
4. Click **Install**, then **Finish** once it is complete.

### Step 3: Verify the Installation

1. Press the **Windows Key**, type cmd, and open the Command Prompt.
2. Run the following command to check that Git is properly recognized:```git --version```

### Step 4: Configure Your User Profile

After installation, you should set your Git username and email so your commits are properly attributed. Run these two commands in your Command Prompt (replacing the example details with your own information):
    
```git config --global user.name "Your Name"```  
```git config --global user.email "your_email@example.com"```

## Professional Git workflow from project initialization to branching, merging, and syncing with GitHub

### 1. Initialize Git Repository

Open Git Bash   
```cd path/to/your/project```

Initialize Git:  
```git init```

Check status:  
```git status```

### 2. Configure Git (First Time Only)

```git config --global user.name "Your Name"```  
```git config --global user.email "your_email@example.com"```

**Verify:**  
```git config --list```

### 3. Create a README File

**Windows:**  
```echo # MyProject > README.md```  
> copy from anywhere or create it manually.

### 4. Add Files to Git

**Add one file:**  
```git add README.md```  

**Add all files:**  
```git add .```  

**Check status:**  
```git status```

### 5. Commit Files

```git commit -m "Initial project setup"```  

**View history:**  
```git log --oneline```

### 6. Connect Local Git to GitHub

Create an empty repository on GitHub.  
Copy the repository URL.  

**Example:**  
```https://github.com/username/MyProject.git```  

**Add remote:**  
```git remote add origin https://github.com/username/MyProject.git```  

**Verify:**  
```git remote -v```

### 7. Rename Default Branch to main

```git branch -M main```  

**Check:**  
```git branch```  

**Output:**  
```main```

### 8. Push main Branch to GitHub

**First push:**  
```git push -u origin main```  

**After this, future pushes only need:**  
```git push```

### 9. Make Changes and Sync main

**Modify any file.**  

**Check status:**  
```git status```  

**Add:**  
```git add .```  

**Commit:**  
```git commit -m "Update project"```  

**Push:**  
```git push```

### 10. Create a Feature Branch

```git checkout -b feature/login```

or

```git switch -c feature/login```

**Check:**  
```git branch```  

**Output:**  

```* feature/login```  
    ```  main```

### 11. Work on Feature Branch

**Modify files.**  

**Add:**  
```git add .```

**Commit:**  
```git commit -m "Add login functionality"```

### 12. Sync Feature Branch with GitHub

**First push:**  
```git push -u origin feature/login```

**Later:**  
```git push```

**GitHub now contains:**  
```main```  
```feature/login```

### 13. Create a Sub-Branch from Feature Branch

**Stay on feature branch:**  
```git checkout feature/login```

**Create sub-branch:**  
```git checkout -b feature/login-ui```  
or  
```git switch -c feature/login-ui```

**Now:**

```main```  
```└── feature/login```  
```    └── feature/login-ui```

### 14. Commit Changes in Sub-Branch

```git add .```  
```git commit -m "Improve login UI"```  

**Push:**  
```git push -u origin feature/login-ui```

### 15. Merge Sub-Branch into Feature Branch

**Switch to feature branch:**  
```git checkout feature/login```

**Merge:**  
```git merge feature/login-ui```

**If there are no conflicts:**  
```Fast-forward```  
    
**Push updated feature branch:**  
```git push```

**Workflow becomes:**

### 16. Merge Feature Branch into main

**Switch to main:**  
```git checkout main```  

**Pull latest version:**  
```git pull origin main```  

**Merge:**  
```git merge feature/login```  

**Push:**   
```git push origin main```

**Now:**

### 17. Delete Merged Branches (Optional)

**Delete local sub-branch:**  
```git branch -d feature/login-ui```  

**Delete local feature branch:**  
```git branch -d feature/login```  

**Delete remote feature branch:**  
```git push origin --delete feature/login```
